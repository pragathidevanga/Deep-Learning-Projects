## **4. Theory: Gradient Descent**

Gradient Descent is an optimization algorithm used to find the values of model parameters that minimize a loss function.

The algorithm starts with initial parameter values. It calculates the gradient of the loss function with respect to the parameters and updates them in the direction that reduces the loss.

The general parameter update equation is:

θ_new = θ_old - α × gradient

Where:

- θ represents a model parameter.
- α represents the learning rate.
- Gradient represents the derivative of the loss function with respect to the parameter.

The process is repeated for multiple iterations until the loss becomes smaller.

## **5. Working of Gradient Descent**

The Gradient Descent process consists of the following steps:

1. Initialize the model parameters.
2. Make predictions using the current parameters.
3. Calculate the loss between actual and predicted values.
4. Calculate the gradients.
5. Update the model parameters.
6. Repeat the process for multiple iterations.
7. Observe whether the loss decreases.

The main purpose is to reduce the prediction error by continuously updating the model parameters.

## **7. Task 1: Creating a Simple Dataset**

A simple dataset is used for this project.

The input variable X represents the input values and y represents the corresponding actual output values.

The dataset approximately follows the relationship:

y = 2x + 1

This makes it suitable for demonstrating how Gradient Descent learns the parameters of a Linear Regression model.

```python
# Creating a simple dataset

X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)

y = np.array([3, 5, 7, 9, 11, 13, 15, 17], dtype=float)

print("Input values (X):")
print(X)

print("\nActual output values (y):")
print(y)
```

## **9. Initializing Model Parameters**

For this project, a simple Linear Regression model is used.

The model equation is:

ŷ = wx + b

Where:

- w = weight
- b = bias
- x = input value
- ŷ = predicted output

Initially, the weight and bias are set to zero.

A learning rate of 0.01 is used, and the model is trained for 1000 iterations.

```python
# Initialize model parameters

w = 0.0
b = 0.0

# Learning rate
learning_rate = 0.01

# Number of iterations
iterations = 1000

print("Initial Weight:", w)
print("Initial Bias:", b)
print("Learning Rate:", learning_rate)
print("Number of Iterations:", iterations)
```

## **10. Prediction Function**

The Linear Regression model predicts the output using the equation:

ŷ = wx + b

The prediction function below calculates the predicted output using the current values of weight and bias.

```python
def predict(X, w, b):
    return w * X + b
```

## **11. Loss Function**

Mean Squared Error (MSE) is used as the loss function.

The formula for Mean Squared Error is:

MSE = (1/n) × Σ(y - ŷ)²

Where:

- y = actual output
- ŷ = predicted output
- n = number of data points

A lower MSE means that the predicted values are closer to the actual values.

Gradient Descent tries to minimize this loss.

```python
def calculate_loss(y, y_pred):
    return np.mean((y - y_pred) ** 2)
```

## **12. Task 2: Implementing Gradient Descent**

Gradient Descent is implemented manually using Python.

For the Linear Regression model, the gradients of the Mean Squared Error with respect to the weight and bias are calculated.

The gradient equations are:

dw = -(2/n) × Σ[x(y - ŷ)]

db = -(2/n) × Σ(y - ŷ)

The parameters are updated using:

w = w - learning_rate × dw

b = b - learning_rate × db

These steps are repeated for multiple iterations so that the loss gradually decreases.

```python
# Reset model parameters

w = 0.0
b = 0.0

learning_rate = 0.01
iterations = 1000

# Lists to store training information
loss_history = []
weight_history = []
bias_history = []

n = len(X)

# Gradient Descent
for i in range(iterations):

    # Step 1: Make predictions
    y_pred = predict(X, w, b)

    # Step 2: Calculate loss
    loss = calculate_loss(y, y_pred)

    # Step 3: Calculate gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Step 4: Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # Store values
    loss_history.append(loss)
    weight_history.append(w)
    bias_history.append(b)

print("Gradient Descent training completed!")
print("Final Weight:", w)
print("Final Bias:", b)
print("Final Loss:", loss_history[-1])
```

## **13. Observing Parameter Updates**

During Gradient Descent, the weight and bias are updated after every iteration.
